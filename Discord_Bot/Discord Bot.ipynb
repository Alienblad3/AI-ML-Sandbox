```python
import speedtest
from datetime import datetime
import os

user = os.getenv('USERNAME')
output = "test.txt"
if not os.path.exists(output):
    open(output, 'w').close()

def test():
    s = speedtest.Speedtest()
    s.get_best_server()
    s.download()
    s.upload()
    res = s.results.dict()
    return res["download"], res["upload"], res['server']["lat"]

with open(output, 'a') as f:
    d, u, p = test()
    now = datetime.now()
    datestr = now.strftime("%Y-%m-%d %H:%M:%S")
    try:
        print(f"[{datestr}] Download: {round(d / 1048576, 1)}MB Upload: {round(u / 1048576, 1)}MB Latency: {p}")
        f.write(f"[{datestr}] Download: {round(d / 1048576, 1)}MB Upload: {round(u / 1048576, 1)}MB Latency: {p}\n")
    except Exception as ex:
        print(f"[{datestr}] An exception of {type(ex)} occurred, {ex.args}")
        f.write(f"[{datestr}] An exception of {type(ex)} occurred, {ex.args}\n")
```

```python
import psutil

def get_network_details():
    # Get network counters
    counters = psutil.net_io_counters(pernic=True)
    # Iterate over network interfaces
    for nic, data in counters.items():
        print(f"Interface: {nic}")
        print(f"  Bytes Sent: {data.bytes_sent}")
        print(f"  Bytes Received: {data.bytes_recv}")
        print(f"  Packets Sent: {data.packets_sent}")
        print(f"  Packets Received: {data.packets_recv}")
        print(f"  Error In: {data.errin}")
        print(f"  Error Out: {data.errout}")
        print(f"  Drop In: {data.dropin}")
        print(f"  Drop Out: {data.dropout}")

if __name__ == "__main__":
    get_network_details()
```

```python
import speedtest
from datetime import datetime
import concurrent.futures

def run_speed_test():
    s = speedtest.Speedtest()
    s.get_best_server()
    s.download()
    s.upload()
    res = s.results.dict()
    return res["download"], res["upload"], res['server']["lat"]

def print_speed_test_result(result):
    d, u, p = result
    now = datetime.now()
    datestr = now.strftime("%Y-%m-%d %H:%M:%S")
    print(f"[{datestr}] Download: {round(d / 1048576, 1)}MB Upload: {round(u / 1048576, 1)}MB Latency: {p}")

with concurrent.futures.ThreadPoolExecutor() as executor:
    future = executor.submit(run_speed_test)
    result = future.result()

print_speed_test_result(result)
```
