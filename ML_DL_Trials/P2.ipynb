```python
import random
import time
def input_generator(size):
    return [random.randint(-1000, 1000)%size for _ in range(size)]
def index(arr):
    left, right = 0, len(arr) - 1
    while left <= right:
        mid = (left + right) // 2
        if arr[mid] == mid:
            return True
        elif arr[mid] < mid:
            left = mid + 1
        else:
            right = mid - 1
    return False
for size in range(1, 10): #Edit Range for Number of Outputs
    arr = input_generator(size)
    start_time2 = time.time()
    print("Input:", arr)
    print("Result:", index(arr))
    end_time2 = time.time()
    time_taken2 = end_time2 - start_time2
    print("Time taken: {:.8f} seconds".format(time_taken2))
```
