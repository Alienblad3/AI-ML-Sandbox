```python
n=int(input()) #No of Pencils
m=int(input()) #No of Erasers
p=int(input()) #Pencils Target
e=int(input()) #Erasers Target
def fact(x):
    if x==0 or x==1:
        return 1
    else:
        return x*fact(x-1)
#to find the number of ways to buy pencils and erasers
pcomb=fact(n)/(fact(n-p)*fact(p))
ecomb=fact(m)/(fact(m-e)*fact(e))
print(pcomb*ecomb)
```
