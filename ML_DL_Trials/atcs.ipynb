```python
import pandas as pd
```

```python
df=pd.read_csv('/home/alienblade/College Stuff/Workspace/data.csv')
df['R-G_Timestamp'] = pd.to_datetime(df['R-G_Timestamp'])
df['G-R_Timestamp'] = pd.to_datetime(df['G-R_Timestamp'])
df['time_diff_seconds'] = (df['G-R_Timestamp'] - df['R-G_Timestamp']).dt.total_seconds()
df['den_diff']=(df['R-G_Desity'] - df['G-R_Density'])
df['traffic_rate']=(df['den_diff'] / df['time_diff_seconds'])
df.info()
```

```python
df['timestamp_unix'] = df['G-R_Timestamp'].astype(int) // 10**9  # Convert nanoseconds to seconds

# Perform linear regression
from sklearn.linear_model import LinearRegression

X = df[['timestamp_unix']]
y = df['traffic_rate']

model = LinearRegression()
model.fit(X, y)

# Predict traffic rates using the model
df['predicted_traffic_rate'] = model.predict(X)

print(df)
```
