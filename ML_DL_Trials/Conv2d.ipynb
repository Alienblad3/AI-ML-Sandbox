```python
import numpy as np
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv2D, MaxPooling2D, Flatten, Dense, Dropout
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras.preprocessing.image import load_img
```

```python
# Define data generators
train_datagen = ImageDataGenerator(rescale=1./255, validation_split=0.2)

# Load and preprocess training data
train_generator = train_datagen.flow_from_directory(
    '/home/alienblade/College Stuff/Code/MLDA/Pet/Master Folder/train',
    target_size=(150, 150),
    batch_size=32,
    class_mode='categorical', 
    subset='training'
)

# Load and preprocess validation data
validation_generator = train_datagen.flow_from_directory(
    '/home/alienblade/College Stuff/Code/MLDA/Pet/Master Folder/valid',
    target_size=(150, 150),
    batch_size=32,
    class_mode='categorical',
    subset='validation'
)
```

```python
model = Sequential()

model.add(Conv2D(32, (3, 3), activation='relu', input_shape=(150, 150, 3)))
model.add(MaxPooling2D((2, 2)))

model.add(Conv2D(64, (3, 3), activation='relu'))
model.add(MaxPooling2D((2, 2)))

model.add(Conv2D(128, (3, 3), activation='relu'))
model.add(MaxPooling2D((2, 2)))

model.add(Conv2D(128, (3, 3), activation='relu'))
model.add(MaxPooling2D((2, 2)))

model.add(Flatten())

model.add(Dense(512, activation='relu'))
model.add(Dropout(0.5))

model.add(Dense(4, activation='softmax'))
```

```python
model.compile(optimizer='adam',
              loss='categorical_crossentropy',
              metrics=['accuracy'])
```

```python
# Calculate steps_per_epoch and validation_steps
steps_per_epoch = train_generator.samples // train_generator.batch_size
validation_steps = validation_generator.samples // validation_generator.batch_size

# Check if there are remaining samples after dividing into batches, if so, add 1 to steps_per_epoch
if train_generator.samples % train_generator.batch_size != 0:
    steps_per_epoch += 1

if validation_generator.samples % validation_generator.batch_size != 0:
    validation_steps += 1

# Model training
model.fit(
    train_generator,
    steps_per_epoch=steps_per_epoch,
    epochs=50,
    validation_data=validation_generator,
    validation_steps=validation_steps
)
```

```python
# Load and preprocess test data
test_datagen = ImageDataGenerator(rescale=1./255)

test_generator = test_datagen.flow_from_directory(
    '/home/alienblade/College Stuff/Code/MLDA/Pet/Master Folder/test',
    target_size=(150, 150),
    batch_size=32,
    class_mode='categorical'
)
```

```python
#Prediction
from tensorflow.keras.preprocessing import image

img_path = '/home/alienblade/College Stuff/Code/MLDA/Pet/Master Folder/test/Sad/014.jpg'
img = image.load_img(img_path, target_size=(150, 150))
img_array = image.img_to_array(img)
img_array = np.expand_dims(img_array, axis=0) / 255.0

predictions = model.predict(img_array)
emotion_labels = ['Angry', 'Happy', 'Other', 'Sad']
predicted_emotion = emotion_labels[np.argmax(predictions)]
print(f'Predicted emotion: {predicted_emotion}')
```
