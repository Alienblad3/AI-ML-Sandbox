Problem 2. You are given a sorted (from smallest to largest) array A of n distinct
integers which can be positive, negative, or zero. You want to decide whether or
not there is an index i such that A[i] = i. Design the fastest algorithm you can for
solving this problem. 

```python
import random
import time
def input_generator(size):
    return [random.randint(-1000, 1000)%size for _ in range(size)]
def index(arr):
    left, right = 0, len(arr) - 1
    while left <= right:
        mid = (left + right) // 2
        if arr[mid] == mid:
            return True
        elif arr[mid] < mid:
            left = mid + 1
        else:
            right = mid - 1
    return False
for size in range(1, 10): #Edit Range for Number of Outputs
    arr = input_generator(size)
    start_time2 = time.time()
    print("Input:", arr)
    print("Result:", index(arr))
    end_time2 = time.time()
    time_taken2 = end_time2 - start_time2
    print("Time taken: {:.8f} seconds".format(time_taken2))
```

Problem 3 You are given an n-by-n grid of distinct numbers. A number is a local minimum if it is smaller than all its neighbors. (A neighbor of a number is one immediately above, below, to the left, or to the right. Most numbers have four neighbors; numbers on the side have three; the four corners have two.) Use the divide-and-conquer algorithm design paradigm to compute a local minimum with only O(n) comparisons between pairs of numbers. 

```python
def find_local_minimum(grid):
    n = len(grid)
    return find_local_minimum_recursive(grid, 0, n-1, n//2)

def find_local_minimum_recursive(grid, start_row, end_row, mid_col):
    min_value = float('inf')
    min_row = -1

    for i in range(start_row, end_row + 1):
        if grid[i][mid_col] < min_value:
            min_value = grid[i][mid_col]
            min_row = i

    left_neighbor = grid[min_row][mid_col - 1] if mid_col > 0 else float('inf')
    right_neighbor = grid[min_row][mid_col + 1] if mid_col < len(grid[0]) - 1 else float('inf')

    if min_value < left_neighbor and min_value < right_neighbor:
        return min_value, min_row, mid_col
    elif min_value > left_neighbor:
        return find_local_minimum_recursive(grid, start_row, end_row, mid_col - (mid_col - start_row) // 2)
    else:
        return find_local_minimum_recursive(grid, start_row, end_row, mid_col + (end_row - mid_col) // 2)

import time
start_time3 = time.time()

grid = [
    [10, 8, 10, 10],
    [14, 13, 12, 11],
    [15, 9, 11, 21],
    [16, 17, 19, 20]
]

min_value, min_row, min_col = find_local_minimum(grid)
print("Local minimum:", min_value, "at index:", (min_row, min_col))
end_time3 = time.time()
time_taken3 = end_time3 - start_time3
print("Time taken: {:.8f} seconds".format(time_taken3))
```
