```python
#Tokenize
sample="""Agent 47 (also known as "The Hitman" and "Mr. 47", or simply by his birth name "47", as well as many aliases) is the playable protagonist of the Hitman series. He is a professional hitman and was the former top agent for the International Contract Agency, until he exposed agency information to the public, ending them permanently. After exposing the ICA and destroying Providence, both 47 and Diana went Freelance and now operate a murder-for-hire service by themselves with no ties to the ICA.

His genetically-engineered and enhanced body and mind (combined with his extensive knowledge, training, and decades worth of experience as a contract killer) made him become the world's deadliest and most efficient assassin.

Agent 47 was created by Dr. Ort-Meyer as his 47th clone in his underground cloning laboratory, disguised under the surface appearance of a sanitarium."""

from nltk.tokenize import  word_tokenize
sample_tokens=word_tokenize(sample)
print(sample_tokens)
print(len(sample_tokens))
```

```python
#FreqDist
from nltk.probability import FreqDist
fd=FreqDist()
for word in sample_tokens:
    fd[word.lower()]+=1
print(fd['hitman'])
```
